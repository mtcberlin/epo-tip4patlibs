## Why this notebook

A client walks in: *"Who in Europe is working on solid-state batteries?"* The first answer is
always a keyword search. It is not wrong — it is **undefendable**. It misses everything phrased
differently and catches everything that merely mentions the term, and neither problem is visible
in the result. You get a list, and no way to prove it is complete.

Someone has already asked your question. This library holds the queries with the decisions
written down: open one, change the country, the time window or the field, and run it. The SQL
stays on the screen, so you can see what was decided for you.

**At the end you have** one query adapted to a question you are actually asked, with its result
table — and a first feel for what a PATSTAT query costs in time.
