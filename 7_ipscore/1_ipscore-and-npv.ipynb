## Step 1 &middot; What IPScore is

Patent analytics usually answers *"who files what, and where"*. **IPScore answers a different
question: what is this patent actually worth, and where is it weak?**

IPScore is an **EPO tool**. You answer **40 questions** grouped into five sections, scoring each
from 1 to 5:

| Section | Questions | Max points | Topic |
|---------|-----------|-----------|-------|
| **A** | 8 | 40 | Legal status |
| **B** | 9 | 45 | Technology |
| **C** | 9 | 45 | Market |
| **D** | 6 | 30 | Finance |
| **E** | 8 | 40 | Strategy |
| **Total** | **40** | **200** | |

The section scores drive two things: a **radar chart** showing the patent's profile at a glance,
and a **risk / opportunity** breakdown. Each question is tagged as a risk driver, an opportunity
driver, or both &mdash; a risk question contributes `-(5 - score) / 4`, an opportunity question
contributes `(score - 1) / 4`. So a low score on a risk question actively pulls the profile down,
which is exactly what a patent attorney would want to see highlighted.

> **For PATLIB staff this is the natural next conversation after a landscape study.** The landscape
> tells a client *where* they stand; IPScore tells them *what to do about a specific patent*.

## Step 2 &middot; The bridge from scores to money &mdash; the OEK factors

This is the clever part of the model. Most of the 40 questions only affect the qualitative
profile &mdash; but **eight of them carry a numeric value** that feeds the financial calculation.
These are the **OEK** questions. A 1&ndash;5 answer is translated into a real economic parameter:

| Question | Meaning | Score 1 &rarr; 5 maps to |
|----------|---------|--------------------------|
| **B5** | Time to first revenue *T* (years) | 5, 2, 1, 0.5, 0 |
| **C2** | Annual market growth *g* | 0.5%, 2.5%, 5%, 8%, 15% |
| **C3** | Commercial lifetime *L* (years) | 0.5, 1, 2, 4, 8 |
| **C6** | Patent market share | 0.5%, 2%, 4%, 6%, 10% |
| **D1** | Substitution factor | 1, 0.75, 0.5, 0.25, 0 |
| **D2** | R&amp;D cost factor | 0.3, 0.15, 0.08, 0.025, 0.005 |
| **D3** | Cost efficiency factor | 1.3, 1.15, 1.0, 0.85, 0.7 |
| **D4** | Investment factor | 1.2, 1.1, 1.0, 0.7, 0.5 |

Note the direction changes: for **B5** a *higher* score means revenue arrives *sooner*
(5 years &rarr; immediately), while for **C3** a higher score means a *longer* commercial life.
The mapping encodes domain judgement, not arithmetic &mdash; which is why it comes from the EPO model
and not from a formula.

## Step 3 &middot; The NPV calculation &mdash; and why you can trust this one

Together with seven financial inputs (turnover, direct and indirect costs, depreciation and
period, sector share, discount rate), the OEK factors drive a ten-year cash-flow projection:

```text
Liquidity[y] = Revenue[y] - Costs[y] - Investments[y]
             + Regained[y] + Efficiency[y] + InvReduction[y]

NPV = Σ  Liquidity[y] × BT/100 / (1 + r)^y      for y = 1 … 10
```

Two non-obvious invariants, both derived by reading the Excel's *Financial Calculations* sheet:

- **Investments and InvestmentReduction are one-time events** in `yFirst` (the first year with
  revenue), **not** annual costs.
- Revenue and the regained/efficiency terms use a **fractional-year factor**, so partial entry and
  exit years are handled properly.

### Why this implementation is credible

This matters more than it might look. The calculation engine was **verified against the IPScore
Excel workbook itself** &mdash; not against its own previous output &mdash; by running the Excel's three
built-in test patents through the code and comparing to Excel's own computed NPVs. All three now
agree to **four or more decimal places**.

That verification found a genuine bug: the average-revenue term summed `depPeriod` years starting
at `yFirst`, while Excel's window is `year > T and year <= T + 1 + depPeriod` &mdash; always one slot
more. It stayed invisible in both demo cases because their commercial lifetime was short enough
that the extra year carried zero revenue, and only surfaced when lifetime is long relative to the
depreciation period.

> **This is the honest reason to show IPScore at a PATLIB workshop:** not "here is a calculator",
> but *"here is a valuation model that was checked against its source of truth, and the check
> caught something real."* Verification is the transferable lesson.

## Step 4 &middot; The assessment tool

Run the cell below to load the interactive IPScore assessment. Work through the sections, or press
**Fill demo** inside the tool to jump straight to a completed example with its radar chart, risk /
opportunity profile, NPV and recommended next steps.

````python
import os, sys, socket, subprocess, time, importlib.util
from pathlib import Path
from IPython.display import display, HTML

PREFIX = os.environ.get("JUPYTERHUB_SERVICE_PREFIX", "/").rstrip("/")
HERE = Path.cwd()
_served = {}


def _start_server():
    """Serve this folder through jupyter-server-proxy.

    Files fetched via Jupyter's /files/ endpoint are delivered with a CSP sandbox
    that disables JavaScript, so an iframe or direct link there shows the layout but
    no working tool. A proxied server is not subject to that restriction.
    """
    if "base" in _served:
        return _served["base"]
    if importlib.util.find_spec("jupyter_server_proxy") is None:
        return None
    sock = socket.socket()
    sock.bind(("", 0))
    port = sock.getsockname()[1]
    sock.close()
    subprocess.Popen([sys.executable, "-m", "http.server", str(port)], cwd=str(HERE),
                     stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
    time.sleep(1.5)
    _served["base"] = f"{PREFIX}/proxy/{port}/"
    return _served["base"]


def open_tool(filename, title):
    """Show a one-click link that opens the tool interactively inside TIP."""
    links = []
    base = _start_server()
    if base:
        links.append(f'<a href="{base}{filename}" target="_blank" style="display:inline-block;'
                     f'background:#be0f05;color:#fff;padding:10px 18px;border-radius:8px;'
                     f'text-decoration:none;font-weight:600;">&#9654; Open {title}</a>')
    # Fallback: force a download, then open it locally — JavaScript runs in your own browser.
    try:
        root = Path(os.environ.get("JUPYTER_SERVER_ROOT") or Path.home()).resolve()
        rel = "/".join(HERE.resolve().relative_to(root).parts)
        dl = f"{PREFIX}/files/{rel}/{filename}?download=1"
        links.append(f'<a href="{dl}" target="_blank" style="margin-left:12px;color:#64748b;'
                     f'font-size:13px;">&#128229; or download it</a>')
    except ValueError:
        pass
    if not base:
        links.append('<div style="margin-top:10px;font-size:12px;color:#94a3b8;">'
                     'jupyter-server-proxy is not available here, so the tool cannot be served '
                     'inside TIP &mdash; download it and open the file in your own browser.</div>')
    display(HTML('<div style="font-family:system-ui,sans-serif;padding:8px 0;">'
                 + "".join(links) + "</div>"))


open_tool("IPscore_EN_Demo.html", "the IPScore assessment")
````
