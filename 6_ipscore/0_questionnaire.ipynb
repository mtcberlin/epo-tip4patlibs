## Why two options, and why neither is a live form inside this notebook

Two designs were tried first, both inside this notebook, and both failed on this platform in a
way that matters: a clickable `ipywidgets` form rendered fine but never synced a single edit
back to the kernel, and sequential `input()` prompts rendered a real text box but the reply
never reached the kernel either - the cell just hung, whether you typed something or accepted
the default.

**Option A** moves the actual data entry out of the notebook entirely, into a standalone HTML
page (`0_questionnaire_tool.html`, built the same way as the Dennemeyer / module 7 tools). It
never talks to a kernel while you fill it in - clicking and typing there is plain browser
JavaScript, nothing round-trips anywhere. When you are done, it hands you a
`questionnaire.json` file to download, and a two-line loader cell here picks that file up from
a fixed folder. That hand-off is a plain file, not a live connection, so it cannot fail the way
widgets or `input()` did. This is the one to use if you are handing the questionnaire to
someone else, or just prefer a form.

**Option B** keeps the values as plain, editable Python literals - the same style as notebook
1's own worked example. No file, no download, no upload: you edit the cell text directly and
run it. This is the one to use if you are comfortable in the notebook and want the fastest
round trip.

```python
from pathlib import Path

import pandas as pd

import ipscore_kit as kit

spec = kit.load_spec()
example = kit.load_worked_example()
p0, f0, sc0 = example["patent"], example["financials"], example["scores"]

print("Previous case (what Step 4 will compare your edits against):")
print(f"  {p0['publication']} - {p0['title']}")
print(f"  {p0['applicant']}")
```

## Step 4 — Save, and see exactly what changed

This is the only cell that writes anything. It saves whatever Option A or Option B captured
above to `0_questionnaire_output/questionnaire.json`, and `ipscore_kit.load_worked_example()`
reads that file from now on - notebooks 2, 3 and 4 need no edit to pick it up. Notebook 1 is
unaffected: it always reproduces the shipped case, on purpose.

The table below is the honest record of what actually changed since the previous save. A row
marked "—" is exactly the previous value, not something you deliberately confirmed.

```python
saved_path = kit.save_questionnaire(patent, financials, scores, financials_note, spec=spec)
print(f"✓ saved to {saved_path.relative_to(Path.cwd())}")
print("  notebooks 2, 3 and 4 will read this file; notebook 1 will not.\n")


def changes(rows):
    return pd.DataFrame([
        {"Field": label, "Previous": old, "New": new,
         "Changed": "changed" if old != new else "—"}
        for label, old, new in rows
    ])


def highlight(row):
    color = "background-color: #fdf2f2" if row["Changed"] == "changed" else ""
    return [color] * len(row)


patent_rows = [(k.replace("_", " ").title(), p0.get(k), patent[k])
               for k in patent if k != "reference"]
fin_rows = [(k.replace("_", " ").title(), getattr(f0, k), financials[k]) for k in financials]
note_rows = [("Financials note", example["financials_note"], financials_note)]
score_rows = [(qid, sc0[qid], scores[qid])
              for qid in sorted(scores, key=lambda x: (x[0], int(x[1:])))]

print("Patent")
display(changes(patent_rows).style.apply(highlight, axis=1).hide(axis="index"))
print("Company figures")
display(changes(fin_rows + note_rows).style.apply(highlight, axis=1).hide(axis="index"))
print("Questionnaire (40 answers)")
display(changes(score_rows).style.apply(highlight, axis=1).hide(axis="index"))

n_changed = sum(1 for qid in scores if scores[qid] != sc0[qid])
print(f"\n{n_changed} of 40 answers changed from the previous case; "
      f"{40 - n_changed} kept their previous value.")
```
