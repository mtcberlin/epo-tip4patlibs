## Setup

Run this cell first to connect to PATSTAT.

```python
from epo.tipdata.patstat import PatstatClient
import pandas as pd
import time

patstat = PatstatClient(env='PROD')

def run_query(query):
    """Execute SQL and return DataFrame with timing info."""
    start = time.time()
    res = patstat.sql_query(query, use_legacy_sql=False)
    elapsed = time.time() - start
    df = pd.DataFrame(res)
    print(f"Query took {elapsed:.2f}s ({len(df)} rows)")
    return df

print("Connected to PATSTAT. Ready to run queries.")
```

## Run MCP-Generated Query

Paste your SQL below and run the cell.

```python
# Paste MCP-generated SQL here
query = """
-- Example: How many patents were filed by German applicants in 2023?
SELECT 
  COUNT(DISTINCT a.appln_id) AS patent_count
FROM tls201_appln a
JOIN tls207_pers_appln pa ON a.appln_id = pa.appln_id
JOIN tls206_person p ON pa.person_id = p.person_id
WHERE 
  pa.applt_seq_nr > 0  -- Only applicants (not inventors)
  AND p.person_ctry_code = 'DE'  -- German applicants
  AND EXTRACT(YEAR FROM a.appln_filing_date) = 2023
"""

df = run_query(query)
df
```
